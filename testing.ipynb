```python
from rdkit import Chem
from rdkit.Chem import Draw
from rdkit.Chem import rdRGroupDecomposition

# 1. Create a molecule with labeled dummy atoms (R1 and R2 groups)
# Isotopes (:1, :2) act as the R-group identifiers
mol = Chem.MolFromSmiles('c1ccccc1[*:1]O[*:2]')

# 2. Relabel the dummy atoms so RDKit renders them as R1, R2 instead of *:1, *:2
rdRGroupDecomposition.RelabelMappedDummies(mol)

# 3. Draw and display the molecule
img = Draw.MolToImage(mol)
img.save('r_group_mol.png')
```
